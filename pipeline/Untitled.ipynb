```python
from sqlalchemy import create_engine
import pandas as pd

engine = create_engine(
    'postgresql+psycopg://root:root@localhost:5432/ny_taxi'
)
```

```python
df_zones = pd.read_csv('taxi_zone_lookup.csv')

df_zones.to_sql(
    name='zones',
    con=engine,
    if_exists='replace'
)
```
